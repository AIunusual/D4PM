```python
"""Train two DDPMs on disjoint halves of the most common MNIST digit.

In Colab, first run `%pip install opacus`, enable a GPU runtime, upload this file,
then run:
    %run /content/mnist_ddpm_dp_model2_colab.py

Each round:
  1) Model 1 trains one epoch with only the true-noise objective.
  2) Model 2 trains one epoch on its disjoint subset using DP-SGD.
  3) Model 1 accumulates a full-subset gradient-distillation pass, then applies
     a plain-SGD update only if a backtracking loss guard accepts it.
"""

import math
import random
import secrets

import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import ConcatDataset, DataLoader, Subset
from torchvision import datasets, transforms
from tqdm.auto import tqdm
from opacus import PrivacyEngine


# Same DDPM architecture and base settings for both models.
SEED = 42
TIMESTEPS = 300
EPOCHS = 100
BATCH_SIZE = 128
LEARNING_RATE = 2e-4
GUIDANCE_WEIGHT = 1.0
KD_NOISE_REPEATS = 4
KD_MAX_BACKTRACK_TRIALS = 8
KD_BACKTRACK_FACTOR = 0.5
SAMPLE_EVERY = 10
DATA_ROOT = "/content/data"

# Model 2 uses DP-SGD. Epsilon=10 is a utility-oriented starting point based
# on published MNIST diffusion experiments; it is weaker privacy than epsilon=1.
# Opacus calculates the Gaussian noise multiplier for this dataset, batch size,
# and number of epochs. The accountant reports the actual epsilon during training.
DP_TARGET_EPSILON = 10.0
DP_TARGET_DELTA = 1e-5
DP_MAX_GRAD_NORM = 1.0
DP_BATCH_SIZE = 64
DP_LEARNING_RATE = 1e-4

# Opacus secure_mode=True requires a compatible torchcsprng installation.
# False is convenient for Colab experiments, but use secure_mode=True and retrain
# from scratch when cryptographically strong randomness is required.
DP_SECURE_RNG = False

random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)


# Combine MNIST train and test, select the most frequent digit, then make
# two disjoint, approximately equal subsets.
transform = transforms.Compose(
    [
        transforms.ToTensor(),
        transforms.Lambda(lambda image: image * 2.0 - 1.0),
    ]
)
mnist_train = datasets.MNIST(
    root=DATA_ROOT,
    train=True,
    download=True,
    transform=transform,
)
mnist_test = datasets.MNIST(
    root=DATA_ROOT,
    train=False,
    download=True,
    transform=transform,
)
all_mnist = ConcatDataset([mnist_train, mnist_test])
all_labels = torch.cat([mnist_train.targets, mnist_test.targets])

digit_counts = torch.bincount(all_labels, minlength=10)
majority_digit = int(digit_counts.argmax().item())
majority_train_count = int((mnist_train.targets == majority_digit).sum().item())
majority_test_count = int((mnist_test.targets == majority_digit).sum().item())

majority_indices = torch.where(all_labels == majority_digit)[0]
split_generator = torch.Generator().manual_seed(SEED)
permuted_indices = majority_indices[
    torch.randperm(len(majority_indices), generator=split_generator)
]
split_point = (len(permuted_indices) + 1) // 2

half_1 = Subset(all_mnist, permuted_indices[:split_point].tolist())
half_2 = Subset(all_mnist, permuted_indices[split_point:].tolist())

print("Combined MNIST digit counts:", digit_counts.tolist())
print(
    f"Selected digit: {majority_digit} "
    f"(train={majority_train_count}, test={majority_test_count}, "
    f"total={len(majority_indices)})"
)
print(f"Model 1 subset: {len(half_1)} images")
print(f"Model 2 subset: {len(half_2)} images")
print("The subsets are disjoint; both original MNIST splits are used for training.")

loader_1 = DataLoader(
    half_1,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=(device.type == "cuda"),
    generator=torch.Generator().manual_seed(SEED + 101),
    drop_last=False,
)
loader_2 = DataLoader(
    half_2,
    batch_size=DP_BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=(device.type == "cuda"),
    generator=torch.Generator().manual_seed(SEED + 102),
    drop_last=False,
)

# Cache Model 1's clean images on CPU. KD uses every image and accumulates
# gradients over micro-batches without changing parameters between batches.
half_1_image_bank = torch.stack(
    [half_1[index][0] for index in range(len(half_1))]
)

# ------------------------------ U-Net model -------------------------------
class SinusoidalPositionEmbeddings(nn.Module):
    """Encode each discrete diffusion timestep as a vector."""

    def __init__(self, dim):
        super().__init__()
        self.dim = dim

    def forward(self, timesteps):
        half_dim = self.dim // 2
        scale = math.log(10000) / (half_dim - 1)
        frequencies = torch.exp(
            torch.arange(half_dim, device=timesteps.device) * -scale
        )
        angles = timesteps.float()[:, None] * frequencies[None, :]
        return torch.cat([angles.sin(), angles.cos()], dim=-1)


class ResBlock(nn.Module):
    def __init__(self, in_channels, out_channels, time_dim):
        super().__init__()
        self.norm1 = nn.GroupNorm(8, in_channels)
        self.conv1 = nn.Conv2d(in_channels, out_channels, 3, padding=1)
        self.time_proj = nn.Linear(time_dim, out_channels)
        self.norm2 = nn.GroupNorm(8, out_channels)
        self.conv2 = nn.Conv2d(out_channels, out_channels, 3, padding=1)
        self.skip = (
            nn.Conv2d(in_channels, out_channels, 1)
            if in_channels != out_channels
            else nn.Identity()
        )

    def forward(self, x, time_embedding):
        h = self.conv1(F.silu(self.norm1(x)))
        h = h + self.time_proj(F.silu(time_embedding))[:, :, None, None]
        h = self.conv2(F.silu(self.norm2(h)))
        return h + self.skip(x)


class SimpleUNet(nn.Module):
    """Small 28x28 U-Net that predicts the noise added to an image."""

    def __init__(self, time_dim=256):
        super().__init__()
        self.time_embedding = nn.Sequential(
            SinusoidalPositionEmbeddings(time_dim),
            nn.Linear(time_dim, time_dim * 2),
            nn.SiLU(),
            nn.Linear(time_dim * 2, time_dim),
        )

        self.input_conv = nn.Conv2d(1, 64, 3, padding=1)
        self.down_block1 = ResBlock(64, 64, time_dim)
        self.downsample1 = nn.Conv2d(64, 128, 4, stride=2, padding=1)
        self.down_block2 = ResBlock(128, 128, time_dim)
        self.downsample2 = nn.Conv2d(128, 128, 4, stride=2, padding=1)

        self.middle1 = ResBlock(128, 128, time_dim)
        self.middle2 = ResBlock(128, 128, time_dim)

        self.up_block1 = ResBlock(256, 128, time_dim)
        self.up_block1b = ResBlock(128, 128, time_dim)
        self.up_block2 = ResBlock(192, 64, time_dim)
        self.up_block2b = ResBlock(64, 64, time_dim)
        self.output_norm = nn.GroupNorm(8, 64)
        self.output_conv = nn.Conv2d(64, 1, 3, padding=1)

    def forward(self, x, timesteps):
        time_embedding = self.time_embedding(timesteps)

        x = self.input_conv(x)
        skip1 = self.down_block1(x, time_embedding)  # 28 x 28
        x = self.downsample1(skip1)
        skip2 = self.down_block2(x, time_embedding)  # 14 x 14
        x = self.downsample2(skip2)  # 7 x 7

        x = self.middle1(x, time_embedding)
        x = self.middle2(x, time_embedding)

        x = F.interpolate(x, size=skip2.shape[-2:], mode="nearest")
        x = self.up_block1(torch.cat([x, skip2], dim=1), time_embedding)
        x = self.up_block1b(x, time_embedding)

        x = F.interpolate(x, size=skip1.shape[-2:], mode="nearest")
        x = self.up_block2(torch.cat([x, skip1], dim=1), time_embedding)
        x = self.up_block2b(x, time_embedding)
        return self.output_conv(F.silu(self.output_norm(x)))


# ----------------------------- DDPM process -------------------------------
def cosine_beta_schedule(timesteps, offset=0.008):
    """Cosine schedule for alpha_bar over a finite number of timesteps."""
    steps = torch.linspace(0, timesteps, timesteps + 1)
    alpha_bar = torch.cos(
        ((steps / timesteps + offset) / (1 + offset)) * math.pi * 0.5
    ) ** 2
    alpha_bar = alpha_bar / alpha_bar[0]
    betas = 1 - alpha_bar[1:] / alpha_bar[:-1]
    return betas.clamp(1e-5, 0.999)


class DDPM:
    def __init__(self, timesteps, device):
        self.timesteps = timesteps
        self.device = device

        self.betas = cosine_beta_schedule(timesteps).to(device)
        self.alphas = 1.0 - self.betas
        self.alpha_bars = torch.cumprod(self.alphas, dim=0)
        self.alpha_bars_prev = F.pad(self.alpha_bars[:-1], (1, 0), value=1.0)
        self.posterior_variance = (
            self.betas
            * (1.0 - self.alpha_bars_prev)
            / (1.0 - self.alpha_bars)
        )

    def training_loss(self, model, clean_images):
        batch_size = clean_images.shape[0]
        t = torch.randint(0, self.timesteps, (batch_size,), device=self.device)
        noise = torch.randn_like(clean_images)

        alpha_bar_t = self.alpha_bars[t][:, None, None, None]
        noisy_images = (
            alpha_bar_t.sqrt() * clean_images
            + (1.0 - alpha_bar_t).sqrt() * noise
        )

        predicted_noise = model(noisy_images, t)
        return F.mse_loss(predicted_noise, noise)

    @torch.no_grad()
    def sample(self, model, num_images=64):
        model.eval()
        images = torch.randn(num_images, 1, 28, 28, device=self.device)

        # Reverse the forward noising process, one DDPM step at a time.
        for step in tqdm(
            reversed(range(self.timesteps)),
            total=self.timesteps,
            desc="Sampling",
        ):
            t = torch.full(
                (num_images,), step, device=self.device, dtype=torch.long
            )
            beta_t = self.betas[t][:, None, None, None]
            alpha_t = self.alphas[t][:, None, None, None]
            alpha_bar_t = self.alpha_bars[t][:, None, None, None]

            predicted_noise = model(images, t)
            mean = images - beta_t / (1.0 - alpha_bar_t).sqrt() * predicted_noise
            mean = mean / alpha_t.sqrt()

            if step > 0:
                variance = self.posterior_variance[t][:, None, None, None]
                images = mean + variance.sqrt() * torch.randn_like(images)
            else:
                images = mean

        return images.clamp(-1.0, 1.0)


# ------------------------------- Train models ------------------------------

# Use different seeds so the two models start from independent weights.
torch.manual_seed(SEED + 1)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED + 1)
model_1 = SimpleUNet().to(device)

torch.manual_seed(SEED + 2)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED + 2)
model_2 = SimpleUNet().to(device)

diffusion = DDPM(timesteps=TIMESTEPS, device=device)
optimizer_1 = torch.optim.AdamW(model_1.parameters(), lr=LEARNING_RATE)
optimizer_1_kd = torch.optim.SGD(
    model_1.parameters(), lr=LEARNING_RATE, momentum=0.0, weight_decay=0.0
)
# Opacus wraps Model 2, its optimizer, and its data loader. It clips each
# example's gradient separately, adds calibrated Gaussian noise, and tracks the
# accumulated privacy loss for the complete 100-epoch run.
optimizer_2 = torch.optim.AdamW(model_2.parameters(), lr=DP_LEARNING_RATE)
privacy_engine = PrivacyEngine(accountant="rdp", secure_mode=DP_SECURE_RNG)

# The model initialization above is reproducible, but the DP noise seed must
# not be the public experiment seed. Use fresh operating-system entropy when
# secure_mode is unavailable; the seed is intentionally never printed/saved.
if not DP_SECURE_RNG:
    private_rng_seed = secrets.randbits(63)
    torch.manual_seed(private_rng_seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(private_rng_seed)

model_2, optimizer_2, loader_2 = privacy_engine.make_private_with_epsilon(
    module=model_2,
    optimizer=optimizer_2,
    data_loader=loader_2,
    target_epsilon=DP_TARGET_EPSILON,
    target_delta=DP_TARGET_DELTA,
    epochs=EPOCHS,
    max_grad_norm=DP_MAX_GRAD_NORM,
    rand_on_empty=True,
)

print(
    "Model 2 DP-SGD configured: "
    f"target epsilon={DP_TARGET_EPSILON}, delta={DP_TARGET_DELTA}, "
    f"noise_multiplier={optimizer_2.noise_multiplier:.4f}, "
    f"max_grad_norm={DP_MAX_GRAD_NORM}, secure_rng={DP_SECURE_RNG}"
)


def get_epsilon_spent():
    """Handle Opacus versions that return epsilon alone or (epsilon, alpha)."""
    result = privacy_engine.get_epsilon(delta=DP_TARGET_DELTA)
    if isinstance(result, tuple):
        result = result[0]
    return float(result)


def train_baseline_epoch(
    model,
    loader,
    optimizer,
    description,
    *,
    report_loss=True,
    clip_batch_gradient=True,
):
    """Train one epoch; Opacus handles per-example clipping for private steps."""
    model.train()
    total_loss = 0.0
    progress = tqdm(loader, desc=description)

    for images, _labels in progress:
        images = images.to(device, non_blocking=True)
        loss = diffusion.training_loss(model, images)

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        if clip_batch_gradient:
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()

        # Do not release raw Model 2 losses: they are data-dependent metrics and
        # are not covered by the DP optimizer's privacy accounting.
        if report_loss:
            total_loss += loss.item()
            progress.set_postfix(loss=f"{loss.item():.4f}")
        else:
            progress.set_postfix(status="DP-SGD")

    return total_loss / len(loader) if report_loss else None


@torch.no_grad()
def evaluate_guard_noise_loss(model, clean_images, timesteps, true_noise):
    """Evaluate the full-subset GT MSE on one fixed corruption bank."""
    was_training = model.training
    model.eval()
    num_images = clean_images.shape[0]
    num_repeats = timesteps.shape[0]
    total_loss = 0.0

    for repeat in range(num_repeats):
        for start in range(0, num_images, BATCH_SIZE):
            end = min(start + BATCH_SIZE, num_images)
            images = clean_images[start:end].to(device, non_blocking=True)
            t = timesteps[repeat, start:end].to(device, non_blocking=True)
            noise = true_noise[repeat, start:end].to(device, non_blocking=True)
            alpha_bar_t = diffusion.alpha_bars[t][:, None, None, None]
            noisy_images = (
                alpha_bar_t.sqrt() * images
                + (1.0 - alpha_bar_t).sqrt() * noise
            )
            predicted_noise = model(noisy_images, t)
            batch_loss = F.mse_loss(predicted_noise, noise)
            total_loss += batch_loss.item() * (end - start) / (
                num_images * num_repeats
            )

    model.train(was_training)
    return total_loss


def train_guided_epoch(student, teacher, clean_images, optimizer):
    """Accumulate full-subset KD gradients and guard one plain-SGD update.

    Each image gets KD_NOISE_REPEATS fixed (t, noise) pairs for this pass.
    Micro-batches limit activation memory, but parameters stay unchanged until
    all examples and corruption pairs have contributed to both gradients.
    Backtracking reuses the exact same bank and accepts only a non-increasing
    full-subset ground-truth noise loss.
    """
    student.train()
    teacher.eval()
    parameters = [parameter for parameter in student.parameters()
                  if parameter.requires_grad]
    num_images = clean_images.shape[0]

    # Fixed for gradient accumulation and every line-search trial in this pass;
    # refreshed on the next epoch to avoid training on one permanent noise draw.
    timestep_bank = torch.randint(
        0, TIMESTEPS, (KD_NOISE_REPEATS, num_images), device="cpu"
    )
    noise_bank = torch.randn(
        KD_NOISE_REPEATS,
        num_images,
        *clean_images.shape[1:],
        dtype=clean_images.dtype,
        device="cpu",
    )

    gt_gradient_sum = [
        torch.zeros_like(parameter, dtype=torch.float32)
        for parameter in parameters
    ]
    guide_gradient_sum = [
        torch.zeros_like(parameter, dtype=torch.float32)
        for parameter in parameters
    ]
    total_gt_loss = 0.0
    total_guide_loss = 0.0
    total_teacher_gt_loss = 0.0
    micro_batches_per_repeat = math.ceil(num_images / BATCH_SIZE)
    progress = tqdm(
        total=KD_NOISE_REPEATS * micro_batches_per_repeat,
        desc="Model 1 | full-subset gradient-KD accumulation",
    )

    optimizer.zero_grad(set_to_none=True)
    for repeat in range(KD_NOISE_REPEATS):
        for start in range(0, num_images, BATCH_SIZE):
            end = min(start + BATCH_SIZE, num_images)
            batch_size = end - start
            images = clean_images[start:end].to(device, non_blocking=True)
            timesteps = timestep_bank[repeat, start:end].to(
                device, non_blocking=True
            )
            true_noise = noise_bank[repeat, start:end].to(
                device, non_blocking=True
            )

            alpha_bar_t = diffusion.alpha_bars[timesteps][:, None, None, None]
            noisy_images = (
                alpha_bar_t.sqrt() * images
                + (1.0 - alpha_bar_t).sqrt() * true_noise
            )

            with torch.no_grad():
                teacher_noise = teacher(noisy_images, timesteps)
                teacher_gt_loss = F.mse_loss(teacher_noise, true_noise)

            student_noise = student(noisy_images, timesteps)
            gt_loss = F.mse_loss(student_noise, true_noise)
            guide_loss = F.mse_loss(student_noise, teacher_noise)
            gt_grads = torch.autograd.grad(
                gt_loss, parameters, retain_graph=True
            )
            guide_grads = torch.autograd.grad(guide_loss, parameters)

            # F.mse_loss averages each micro-batch. Weight by its share of the
            # full image set and by the number of corruption repeats.
            contribution_weight = batch_size / (
                num_images * KD_NOISE_REPEATS
            )
            total_gt_loss += gt_loss.detach().item() * contribution_weight
            total_guide_loss += guide_loss.detach().item() * contribution_weight
            total_teacher_gt_loss += (
                teacher_gt_loss.detach().item() * contribution_weight
            )
            for index, (gt_grad, guide_grad) in enumerate(
                zip(gt_grads, guide_grads)
            ):
                gt_gradient_sum[index].add_(
                    gt_grad.detach().float(), alpha=contribution_weight
                )
                guide_gradient_sum[index].add_(
                    guide_grad.detach().float(), alpha=contribution_weight
                )

            progress.set_postfix(
                gt=f"{gt_loss.detach().item():.4f}",
                guide=f"{guide_loss.detach().item():.4f}",
                repeat=f"{repeat + 1}/{KD_NOISE_REPEATS}",
            )
            progress.update(1)
            del gt_grads, guide_grads, student_noise, gt_loss, guide_loss
            del teacher_gt_loss

    progress.close()

    dot_product = torch.zeros((), device=device)
    gt_norm_sq = torch.zeros((), device=device)
    guide_norm_sq = torch.zeros((), device=device)
    for gt_grad, guide_grad in zip(gt_gradient_sum, guide_gradient_sum):
        dot_product += torch.sum(gt_grad * guide_grad)
        gt_norm_sq += torch.sum(gt_grad.square())
        guide_norm_sq += torch.sum(guide_grad.square())

    cosine = dot_product / (
        gt_norm_sq.sqrt() * guide_norm_sq.sqrt() + 1e-12
    )
    cosine_gate = cosine.clamp(min=0.0, max=1.0)
    gt_grad_norm = gt_norm_sq.sqrt()
    guide_grad_norm = guide_norm_sq.sqrt()
    norm_ratio = gt_grad_norm / (guide_grad_norm + 1e-12)
    coefficient = (cosine_gate * norm_ratio).clamp(max=1.0).detach()
    cosine_value = cosine.detach().item()
    coefficient_value = coefficient.item()

    # This pre-update loss is evaluated on the same full-subset corruption bank
    # that produced the accumulated gradients.
    guard_loss_before = evaluate_guard_noise_loss(
        student, clean_images, timestep_bank, noise_bank
    )
    guard_loss_after = guard_loss_before
    accepted_lr = 0.0
    backtracks = 0
    update_accepted = False
    last_trial_loss = None

    if cosine_value <= 0.0:
        print(
            "KD update skipped before line search: full-subset cosine is "
            f"non-positive (cosine={cosine_value:.4f})."
        )
    elif coefficient_value <= 0.0:
        print(
            "KD update skipped before line search: guidance coefficient is "
            f"zero (cosine={cosine_value:.4f}, coefficient={coefficient_value:.4f})."
        )
    else:
        for parameter, guide_grad in zip(parameters, guide_gradient_sum):
            parameter.grad = (
                GUIDANCE_WEIGHT
                * coefficient.to(dtype=parameter.dtype)
                * guide_grad.to(dtype=parameter.dtype)
            )
        # Keep the raw SGD step bounded; backtracking checks the actual loss.
        torch.nn.utils.clip_grad_norm_(parameters, max_norm=1.0)
        original_parameters = [parameter.detach().clone() for parameter in parameters]

        for trial in range(KD_MAX_BACKTRACK_TRIALS):
            trial_lr = LEARNING_RATE * (KD_BACKTRACK_FACTOR ** trial)
            for group in optimizer.param_groups:
                group["lr"] = trial_lr
            optimizer.step()

            last_trial_loss = evaluate_guard_noise_loss(
                student, clean_images, timestep_bank, noise_bank
            )
            if last_trial_loss <= guard_loss_before:
                update_accepted = True
                accepted_lr = trial_lr
                backtracks = trial
                guard_loss_after = last_trial_loss
                print(
                    "KD update accepted: full-subset guard GT MSE "
                    f"{guard_loss_before:.6f} -> {guard_loss_after:.6f}; "
                    f"SGD lr={accepted_lr:.2e}, backtracks={backtracks}."
                )
                break

            with torch.no_grad():
                for parameter, original in zip(parameters, original_parameters):
                    parameter.copy_(original)

        for group in optimizer.param_groups:
            group["lr"] = LEARNING_RATE

        if not update_accepted:
            print(
                "KD update skipped: line search exhausted "
                f"{KD_MAX_BACKTRACK_TRIALS} SGD trials; full-subset guard GT "
                f"MSE rose on every trial (before={guard_loss_before:.6f}, "
                f"last candidate={last_trial_loss:.6f})."
            )

    optimizer.zero_grad(set_to_none=True)
    return {
        "gt_loss": total_gt_loss,
        "guide_loss": total_guide_loss,
        "teacher_gt_loss": total_teacher_gt_loss,
        "global_cosine": cosine_value,
        "coefficient": coefficient_value,
        "guard_gt_before": guard_loss_before,
        "guard_gt_after": guard_loss_after,
        "update_accepted": update_accepted,
        "accepted_lr": accepted_lr,
        "backtracks": backtracks,
        "noise_repeats": KD_NOISE_REPEATS,
    }


@torch.no_grad()
def show_samples(model, title, seed):
    # Reuse the same random sampling stream for before/after comparisons.
    cpu_rng_state = torch.get_rng_state()
    cuda_rng_state = torch.cuda.get_rng_state_all() if torch.cuda.is_available() else None

    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    samples = diffusion.sample(model, num_images=32)
    samples = (samples.cpu() + 1.0) / 2.0

    torch.set_rng_state(cpu_rng_state)
    if cuda_rng_state is not None:
        torch.cuda.set_rng_state_all(cuda_rng_state)

    fig, axes = plt.subplots(4, 8, figsize=(8, 4))
    fig.suptitle(title)
    for index, ax in enumerate(axes.flat):
        ax.imshow(samples[index, 0], cmap="gray", vmin=0.0, vmax=1.0)
        ax.axis("off")
    plt.tight_layout()
    plt.show()


def save_checkpoint(model, optimizer, model_name, epoch):
    path = (
        f"/content/mnist_digit_{majority_digit}_{model_name}_"
        f"epoch_{epoch:03d}.pth"
    )
    model_to_save = getattr(model, "_module", model)
    checkpoint = {
        "model_state_dict": model_to_save.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "selected_digit": majority_digit,
        "training_epoch": epoch,
        "timesteps": TIMESTEPS,
        "batch_size": DP_BATCH_SIZE if model_name == "model2_dp" else BATCH_SIZE,
        "learning_rate": DP_LEARNING_RATE if model_name == "model2_dp" else LEARNING_RATE,
        "guidance_weight": GUIDANCE_WEIGHT,
        "optimizer_class": type(optimizer).__name__,
        "kd_optimizer": "SGD (no momentum, no weight decay)",
        "kd_noise_repeats": KD_NOISE_REPEATS,
        "kd_max_backtrack_trials": KD_MAX_BACKTRACK_TRIALS,
        "model_name": model_name,
    }
    if model_name == "model2_dp":
        checkpoint.update(
            {
                "dp_target_epsilon": DP_TARGET_EPSILON,
                "dp_target_delta": DP_TARGET_DELTA,
                "dp_epsilon_spent": get_epsilon_spent(),
                "dp_max_grad_norm": DP_MAX_GRAD_NORM,
                "dp_noise_multiplier": optimizer.noise_multiplier,
                "dp_secure_rng": DP_SECURE_RNG,
                "dp_accountant": "rdp",
            }
        )
    torch.save(checkpoint, path)
    print(f"Saved: {path}")


for epoch in range(1, EPOCHS + 1):
    # First establish Model 1's pure-DDPM baseline for this round.
    pure_loss_1 = train_baseline_epoch(
        model_1,
        loader_1,
        optimizer_1,
        f"Model 1 pure DDPM | Epoch {epoch}/{EPOCHS}",
    )

    # Train the teacher on its own, disjoint half with per-example DP-SGD.
    # Do not publish raw data-dependent training losses.
    train_baseline_epoch(
        model_2,
        loader_2,
        optimizer_2,
        f"Model 2 DP-SGD | Epoch {epoch}/{EPOCHS}",
        report_loss=False,
        clip_batch_gradient=False,
    )
    model_2_epsilon = get_epsilon_spent()

    if epoch % SAMPLE_EVERY == 0:
        print(f"\nEpoch {epoch}: Model 1 before distillation")
        save_checkpoint(model_1, optimizer_1, "model1_pre_kd", epoch)
        show_samples(
            model_1,
            f"Model 1 after pure DDPM, before KD | epoch {epoch}",
            seed=SEED + 1000 + epoch,
        )

    # Accumulate gradients over all of Model 1's images before a single guarded
    # SGD update. Micro-batches are only for limiting activation memory.
    kd_metrics = train_guided_epoch(
        model_1,
        model_2,
        half_1_image_bank,
        optimizer_1_kd,
    )

    print(
        f"Epoch {epoch}/{EPOCHS} | "
        f"Model 1 pure loss={pure_loss_1:.4f} | "
        f"Model 2 DP epsilon={model_2_epsilon:.3f}/"
        f"{DP_TARGET_EPSILON:.1f} (delta={DP_TARGET_DELTA:.1e}) | "
        f"Half1 fixed-bank GT MSE | "
        f"Model1 pre-KD={kd_metrics['guard_gt_before']:.4f}, "
        f"Model2 teacher={kd_metrics['teacher_gt_loss']:.4f}, "
        f"Model1 post-KD={kd_metrics['guard_gt_after']:.4f} | "
        f"KD loss={kd_metrics['guide_loss']:.4f} | "
        f"global cosine={kd_metrics['global_cosine']:.3f} | "
        f"coefficient={kd_metrics['coefficient']:.3f} | "
        f"KD update={'accepted' if kd_metrics['update_accepted'] else 'skipped'} | "
        f"SGD lr={kd_metrics['accepted_lr']:.2e} | "
        f"backtracks={kd_metrics['backtracks']} | "
        f"noise repeats={kd_metrics['noise_repeats']}"
    )

    if epoch % SAMPLE_EVERY == 0:
        print(f"\nEpoch {epoch}: after distillation")
        show_samples(
            model_1,
            f"Model 1 after gradient KD | epoch {epoch}",
            seed=SEED + 1000 + epoch,
        )
        show_samples(
            model_2,
            f"Model 2 DP-SGD | epoch {epoch} | "
            f"eps={model_2_epsilon:.2f}, delta={DP_TARGET_DELTA:.0e}",
            seed=SEED + 1000 + epoch,
        )
        save_checkpoint(model_1, optimizer_1_kd, "model1_post_kd", epoch)
        save_checkpoint(model_2, optimizer_2, "model2_dp", epoch)

if EPOCHS % SAMPLE_EVERY != 0:
    save_checkpoint(model_1, optimizer_1, "model1_pre_kd", EPOCHS)
    save_checkpoint(model_1, optimizer_1_kd, "model1_post_kd", EPOCHS)
    save_checkpoint(model_2, optimizer_2, "model2_dp", EPOCHS)
```
